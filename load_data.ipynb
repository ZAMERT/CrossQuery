```python
import ir_datasets

subforums = ["android", "programmers", "unix"]

for subforum in subforums:
    dataset = ir_datasets.load(f"beir/cqadupstack/{subforum}")

    docs = list(dataset.docs_iter())
    queries = list(dataset.queries_iter())
    qrels = list(dataset.qrels_iter())

    print(subforum, len(docs), len(queries), len(qrels))
```
